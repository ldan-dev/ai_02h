# Clustering

Es la tarea de identificar instancias similares y organizarlas por grupos, sin disponer previamente de sus etiquetas de clase


-----


## ¿Dónde se usa?

- Segmentación de clientes
- Análisis de datos
- Como técnica de reducción de dimensionalidades
- Detección de anomalías (outliers)
- Para aprendizaje semi-supervisado
- Motores de búsqueda
- segmentación de imágenes


-------

## Clustering vs Clasificación

| Clusterin                                  | Clase                                       |
|--------------------------------------------|---------------------------------------------|
| Grupo inferido a partir de la estructura de las características  | Categoría asociada a la salida objetivo en un problema supervisado |
|      $G_j \subseteq X$   ($\subseteq$: "subconjunto de")        |         $y^{(i)} = \in \{C_1, \dots, C_k\} $  ($\in$ "pertenece a")           |
| Ej: Grupo A, Grupo B,...      | Ej: perro, gato, ave       |
| Los grupos no vienen acompañados de un significado semántico | La categoría es conocida durante el entrenamiento |



--------

## ¿Qué es un cluster?

No existe una definición única, depende del contexto.
- Algunos algoritmos buscan por instancias centradas alrededor de cierto punto, llamado *centroide*
- Otros buscan en regiones densas continuas que pueden tomar cualquier forma



-------


## Las 2 familias de clustering



| Métodos particionales                                  | Métodos jerárquicos                                       |
|--------------------------------------------|---------------------------------------------|
|   |  |






---------


### Métodos particionales

- Buscan directamente una partición:

$$
C = \{C_1, \dots, C_k \}
$$

y la mejoran mediante un criterio objetivo

- Normalmente cada clúster se resume mediante un prototipo

- Ejemplos: k-means, k-medoids, k-medians

----

### Métodos jerárquicos

Construyen una organización anidada de clusters, usualmente representada mediante un **dendograma**

Pueden ser:
- aglomerativos: de puntos individuales hacia grupos mayores
- divisivos: de un grupo global hacia subdivisiones
